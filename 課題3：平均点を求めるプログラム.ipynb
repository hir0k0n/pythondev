```python
total_score = 0
total_people = 0

while True:
    score = int(input("点数を入力してください："))

    if score == -1:
        break
    
    elif score <= -2:
        pass
    
    else:
        total_score += score
        total_people += 1

average_score = round(total_score / total_people, 1)

print(f"{total_people}人のテストの平均点は{average_score}点です")
```
