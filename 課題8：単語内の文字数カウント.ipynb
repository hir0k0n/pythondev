```python
#@title 英単語を格納するリストの作成
list_word = list()
```

```python
#@title アルファベットの辞書の作成
dict_alphabet = dict.fromkeys(list("abcdefghijklmnopqrstuvwxyz"), 0)
```

```python
#@title 英単語の入力受付けと格納処理
while True:
    word = input("英単語を入力してください：")
    if word == "":
        break
    else:
        list_word.append(word)
        for i in word:
            dict_alphabet[i] += 1
```

```python
#@title 出力処理
list_word.sort()
print(list_word)

for key, value in dict_alphabet.items():
    if value == 0:
        continue
    else:
        print(f"{key}が{value}個ありました")
```
