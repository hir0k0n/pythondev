```python
class Baseball_Team:
        
    def __init__(self, name, win, lose, draw):
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw
    
    def calc_win_rate(self):
        return '{:.03f}'.format(self.win / (self.win + self.lose))
        
    def show_team_result(self):
        format_name = str(self.name).ljust(9)
        format_win = str(self.win).rjust(3)
        format_lose = str(self.lose).rjust(4)
        format_draw = str(self.draw).rjust(4)
        print(f"{format_name} {format_win} {format_lose} {format_draw} {self.calc_win_rate()}")

    @classmethod
    def show_title(self):
        title_name = "team".ljust(8)
        title_win = "win".rjust(4)
        title_lose = "lose".rjust(4)
        title_draw = "draw".rjust(4)
        title_rate = "rate".rjust(4)
        print(f"{title_name} {title_win} {title_lose} {title_draw} {title_rate}")
```

```python

Baseball_Team.show_title()
Baseball_Team("Archers", 77, 64, 2).show_team_result()
Baseball_Team("Meteors", 71, 69, 3).show_team_result()
Baseball_Team("Cheetahs", 69, 68, 6).show_team_result()
Baseball_Team("Sharks", 70, 70, 3).show_team_result()
Baseball_Team("Pythons", 68, 73, 2).show_team_result()
Baseball_Team("Owls", 59, 82, 2).show_team_result()
```
